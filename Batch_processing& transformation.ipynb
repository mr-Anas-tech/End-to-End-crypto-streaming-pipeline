```python
storage_account_name = "awcriptoprojectstoracc"
storage_account_key = dbutils.secrets.get(scope = 'crypto-data-key', key = 'storagekey').strip()

spark.conf.set(f"fs.azure.account.key.{storage_account_name}.dfs.core.windows.net", storage_account_key)
display(dbutils.fs.ls(f"abfss://bronze@{storage_account_name}.dfs.core.windows.net/"))
```

```python
file_path = f"abfss://bronze@{storage_account_name}.dfs.core.windows.net/historical_crypto_data/data_0acd7802-4cfe-49de-997a-61b375e651b4_f0475e3b-fc47-49f5-bbb7-8f54c784dffc.csv"
df = spark.read.format("csv").option("header", "false").load(file_path)
display(df.limit(100))
```

```python
df_historical_rnm = (
    df.withColumnRenamed("_c0", "trade_id")
    .withColumnRenamed("_c1", "price")
    .withColumnRenamed("_c2", "qty")
    .withColumnRenamed("_c3", "quote_qty")
    .withColumnRenamed("_c4", "time")
    .withColumnRenamed("_c5", "is_buyer_maker")
    .withColumnRenamed("_c6", "is_best_match")
)

display(df_historical_rnm.limit(100))
```

```python
df_historical_rnm.write\
    .format("delta")\
        .mode("append")\
            .saveAsTable("Crypto_project_cat.default.historical_crypto_data")
```
